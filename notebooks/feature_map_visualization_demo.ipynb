# Quantum Feature Map Visualization Demo

This short tutorial shows how classical 2D data can be embedded into quantum states using feature maps. We look at kernel matrices to see how similarity changes after the embedding.

*The notebook keeps plots in SVG format so they stay lightweight and diff-friendly.*

```python
%config InlineBackend.figure_formats = ['svg']
```

## 1. Load helpers
We import the dataset generators, feature map builders, and embedding utilities from the `app` package.

```python
from app.dataset import make_grid_data
from app.feature_maps import build_angle_encoding_feature_map, build_zz_feature_map
from app.embedding import compute_statevectors, compute_kernel_matrix
from app.visualizer import plot_embedding_kernel, plot_2d_points
import matplotlib.pyplot as plt
%config InlineBackend.figure_formats = ['svg']
```

## 2. Create a simple dataset
Grids are easy to visualize and show how evenly spaced points move in feature space.

```python
X = make_grid_data(4)
print(f"Generated {len(X)} points")
plot_2d_points(X, 'examples/notebook_grid.svg', title='Grid dataset preview')
```

## 3. Angle encoding feature map
Single-qubit rotations encode each feature value. We compute the kernel matrix to see similarity after encoding.

```python
angle_map = build_angle_encoding_feature_map(num_qubits=2)
angle_states = compute_statevectors(angle_map, X)
angle_kernel = compute_kernel_matrix(angle_states)
plot_embedding_kernel(angle_kernel, 'examples/notebook_angle_kernel.svg')
angle_kernel[:3, :3]
```

## 4. ZZFeatureMap
The ZZ feature map adds entanglement, letting features interact. The kernel can look quite different.

```python
zz_map = build_zz_feature_map(num_qubits=2, reps=2)
zz_states = compute_statevectors(zz_map, X)
zz_kernel = compute_kernel_matrix(zz_states)
plot_embedding_kernel(zz_kernel, 'examples/notebook_zz_kernel.svg')
zz_kernel[:3, :3]
```

## 5. Takeaways
- Feature maps are quantum circuits that turn numbers into quantum states.
- Kernel matrices capture how similar embedded states are.
- Different feature maps reshape the geometry of the data, which can help machine learning models discover new structure.
